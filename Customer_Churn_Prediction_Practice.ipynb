```python
df.dropna(inplace=True)
```

```python
df['TotalCharges']=pd.to_numeric(df['TotalCharges'],errors='coerce')
```

```python
df['TotalCharges']=df['TotalCharges'].fillna(df['MonthlyCharges']*df['tenure'])
#here the fillna is used to multiply the monthlycharges*tenure that gives totalchRGES bcoz the totalcharges has some NaN values
```

```python
df.drop(columns=['customerID'],inplace=True)
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
```

```python
for col in df.columns:
  if df[col].dtype=='object':
    df[col]=le.fit_transform(df[col])
```

```python
df.info()
```

```python
#Model Building
x=df.drop(columns=['Churn'])
y=df['Churn']
```

```python
x
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,train_size=0.8,random_state=42)
```

```python
print(x_train.shape,x_test.shape,y_train.shape,y_test.shape)
```

```python
from sklearn.tree import DecisionTreeClassifier #we imported decision tree from sklearn library
dt=DecisionTreeClassifier() #we defined the model as dt
```

```python
dt.fit(x_train,y_train) #we learnt the training data for the dt model
```

```python
y_pred_dt=dt.predict(x_test)#predictions made by decision tree using x_test
```

```python
y_pred_dt
```

```python
from sklearn.metrics import accuracy_score,classification_report
```

```python
dt_accuracy=accuracy_score(y_test,y_pred_dt)
print("Accuracy within decisiontree",dt_accuracy)
```

```python
from sklearn.ensemble import RandomForestClassifier
rf=RandomForestClassifier()
```

```python
rf.fit(x_train,y_train)
```

```python
y_pred_rf=rf.predict(x_test)
```

```python
y_pred_rf
```

```python
rf_accuracy=accuracy_score(y_test,y_pred_rf)
print("Accuracy with random forest model",rf_accuracy)
```
